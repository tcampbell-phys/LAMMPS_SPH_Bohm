```python
def write_intermittent_damping_dump_outputs(dump_path,timestep_len,N_stage,N_step_damp,N_step_NVE,N_step_output,N_step_output_freq,damping_coefficient):
  """Script to write run structure of LAMMPS input file. Tailored to writing intermittent stages of damping and NVE.
  Inputs:
  dump_path - path for dump file output
  timestep_len - length of timestep
  N_stage - number of NVE-damp cycles to run
  N_step_damp - number of timesteps per damp
  N_step_NVE - number of timesteps per NVE
  N_step_output - number of timesteps over which to dump atom properties
  N_step_output_freq - frequency of dump
  damping_coefficient - damping strength
  """

  count = 0

  # initial commands
  print("variable damping_strength equal "+str(damping_coefficient)+'\n')
  print("fix all_nve all nve\n")
  print("variable equi_time_steps equal "+str(N_step_damp)+"\n")
  print("variable buff_time_steps equal "+str(N_step_NVE)+"\n")
  print("variable CoM_check_time_steps equal "+str(N_step_output)+"\n")
  print("variable dump_freq equal "+str(N_step_output_freq)+"\n")
  print("variable time_step_length equal "+str(timestep_len)+"\n")
  print("timestep ${time_step_length}\n")


  for i in range(N_stage):
    print('### stage '+str(i))
    print("run ${buff_time_steps}\n")
    print("fix damp_all"+str(i)+" all viscous ${damping_strength}\n")
    print("run ${equi_time_steps}\n")
    print("unfix damp_all"+str(i)+"\n")
    print("dump "+str(count)+" ion xyz ${dump_freq} "+dump_path+"${outputdir}/ion_xyz_data_check"+str(i)+".dump\n")
    count +=1
    print("dump "+str(count)+" electron custom ${dump_freq} "+dump_path+"${outputdir}/ele_xyzh_data_check"+str(i)+".dump id x y z width_SPH x_COM y_COM z_COM vx vy vz\n")
    count +=1
    print("run ${CoM_check_time_steps}\n")
    print("undump "+str(count-2)+"\n")
    print("undump "+str(count-1)+"\n")

  print('quit 0')

def write_damping_dump_outputs(dump_path,timestep_len,N_stage,N_step_damp,N_step_output,N_step_output_freq,damping_coefficient):
  """Script to write run structure of LAMMPS input file. Tailored to writing intermittent stages of damping and NVE.
  Inputs:
  dump_path - path for dump file output
  timestep_len - length of timestep
  N_stage - number of NVE-damp cycles to run
  N_step_damp - number of timesteps per damp
  N_step_output - number of timesteps over which to dump atom properties
  N_step_output_freq - frequency of dump
  damping_coefficient - damping strength
  """

  count = 0

  # initial commands
  print("variable damping_strength equal "+str(damping_coefficient)+'\n')
  print("fix all_nve all nve\n")
  print("variable equi_time_steps equal "+str(N_step_damp)+"\n")
  print("variable CoM_check_time_steps equal "+str(N_step_output)+"\n")
  print("variable dump_freq equal "+str(N_step_output_freq)+"\n")
  print("variable time_step_length equal "+str(timestep_len)+"\n")
  print("timestep ${time_step_length}\n")
  print("fix damp_all all viscous ${damping_strength}\n")


  for i in range(N_stage):
    print('### stage '+str(i))
    print("run ${equi_time_steps}\n")
    print("dump "+str(count)+" ion xyz ${dump_freq} "+dump_path+"${outputdir}/ion_xyz_data_check"+str(i)+".dump\n")
    count +=1
    print("dump "+str(count)+" electron custom ${dump_freq} "+dump_path+"${outputdir}/ele_xyzh_data_check"+str(i)+".dump id x y z width_SPH x_COM y_COM z_COM vx vy vz\n")
    count +=1
    print("run ${CoM_check_time_steps}\n")
    print("undump "+str(count-2)+"\n")
    print("undump "+str(count-1)+"\n")

  print('quit 0')


dump_path = '/work4/clf/scarf734/LAMMPS/SPH_Bohm/ground/'
timestep_len = 0.0005e-3
N_stage = 36
N_step_damp = 4e4
N_step_NVE = 4e4
N_step_output = 25
N_step_output_freq = 1
damping_coefficient = 1e-4

write_intermittent_damping_dump_outputs(dump_path,timestep_len,N_stage,N_step_damp,N_step_NVE,N_step_output,N_step_output_freq,damping_coefficient)
```
